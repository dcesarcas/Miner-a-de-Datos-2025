```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import train_test_split
from google.colab import drive
drive.mount("/content/drive")
```

```python
df = pd.read_excel("/content/drive/MyDrive/base-de-datos-violencia-intrafamiliar-ano-2024_v3 (2).xlsx")
```

```python
data = df
```

```python
data.head()
```

```python
data['etnica'] = data['AGR_GURPET'].apply(lambda x: 1 if x == 4 else 0)
```

```python
data =data.sample(frac=1, random_state=42).reset_index(drop='TRUE')
```

```python
x = data[['VIC_DISC','VIC_ESCOLARIDAD','VIC_SEXO','HEC_AREA']]
x = x.fillna(0)
y = data['etnica']
```

```python
x_train, x_test, y_train ,y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
model = Sequential()
model.add(Dense(8, input_dim=4, activation='relu'))
model.add(Dense(4,activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
model.fit(x_train, y_train, epochs=50, batch_size=64, validation_data=(x_test, y_test))
```

```python
loss, accuracy = model.evaluate(x_test, y_test)
print(f'Loss: {loss}, Accuracy: {accuracy}')
```

```python
perfil1 = np.array([[1,39,2,2]])
res = model.predict(perfil1)
print(res)
```
